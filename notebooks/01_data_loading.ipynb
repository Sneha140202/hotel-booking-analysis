## Key Column Definitions

Below is a reference guide for the most important columns in the dataset.

---

### 🏨 Hotel & Booking Identity

| Column | Type | Description |
|--------|------|-------------|
| `hotel` | str | Type of hotel — **Resort Hotel** or **City Hotel** |
| `is_canceled` | int (0/1) | Target variable — **1** = booking was cancelled, **0** = not cancelled |
| `reservation_status` | str | Final status: `Check-Out`, `Canceled`, or `No-Show` |
| `reservation_status_date` | str | Date when the last status was set |

---

### 📅 Arrival & Stay Details

| Column | Type | Description |
|--------|------|-------------|
| `lead_time` | int | Number of days between booking date and arrival date |
| `arrival_date_year` | int | Year of arrival |
| `arrival_date_month` | str | Month of arrival (e.g., `July`) |
| `arrival_date_week_number` | int | ISO week number of the arrival date |
| `arrival_date_day_of_month` | int | Day of the month of arrival |
| `stays_in_weekend_nights` | int | Number of weekend nights (Sat/Sun) included in the stay |
| `stays_in_week_nights` | int | Number of weekday nights (Mon–Fri) included in the stay |

---

### 👥 Guest Profile

| Column | Type | Description |
|--------|------|-------------|
| `adults` | int | Number of adults |
| `children` | float | Number of children |
| `babies` | int | Number of babies |
| `country` | str | Country of origin of the guest (ISO 3166 code) |
| `customer_type` | str | Booking type: `Transient`, `Contract`, `Group`, `Transient-Party` |
| `is_repeated_guest` | int (0/1) | **1** = guest has stayed at the hotel before |
| `previous_cancellations` | int | Number of previous bookings the guest cancelled |
| `previous_bookings_not_canceled` | int | Number of previous bookings the guest honoured |

---

### 💰 Pricing & Revenue

| Column | Type | Description |
|--------|------|-------------|
| `adr` | float | **Average Daily Rate** — average revenue per occupied room per day (core pricing KPI) |
| `deposit_type` | str | Deposit policy: `No Deposit`, `Non Refund`, `Refundable` |
| `required_car_parking_spaces` | int | Number of parking spaces requested by the guest |
| `total_of_special_requests` | int | Number of special requests made (e.g., high floor, twin bed) |

---

### 🏷️ Booking Channel & Room

| Column | Type | Description |
|--------|------|-------------|
| `market_segment` | str | Booking channel segment (e.g., `Direct`, `Online TA`, `Corporate`) |
| `distribution_channel` | str | Distribution channel used (e.g., `Direct`, `GDS`, `TA/TO`) |
| `reserved_room_type` | str | Room type originally reserved (anonymised codes: A–H) |
| `assigned_room_type` | str | Room type actually assigned at check-in |
| `booking_changes` | int | Number of changes made to the booking before check-in |
| `agent` | float | ID of the travel agency that made the booking (`NULL` = no agent) |
| `company` | float | ID of the company that made the booking (`NULL` = no company) |

---

### 🍽️ Services

| Column | Type | Description |
|--------|------|-------------|
| `meal` | str | Meal package booked: `BB` (Bed & Breakfast), `HB` (Half Board), `FB` (Full Board), `SC`/`Undefined` (no meal) |
| `days_in_waiting_list` | int | Number of days the booking waited on a waitlist before being confirmed |

---

> **Key target for modelling:** `is_canceled`  
> **Key metric for pricing analysis:** `adr` (Average Daily Rate)
